```python
import pandas as pd

def add_to_dataframe_hagstofudateformat(
    df: pd.DataFrame,
    csv_path: str,
    incoming_month: str,
    incoming_data: str,
    new_column_name: str,
    datetimeformat="%d.%m.%Y"
) -> pd.DataFrame:
    """
    Adds a monthly-aligned column to an existing DataFrame that has a 'time' column.

    Parameters
    ----------
    df : pd.DataFrame
        Target DataFrame).
    csv_path : str
        Path to the CSV with monthly dates + values to merge in.
    incoming_month : str
        Column name in the CSV representing the month.
    incoming_data : str
        Column name in the CSV representing the values.
    new_column_name : str
        Name for the new column to be added to df.

    Returns
    -------
    pd.DataFrame
        A new DataFrame with the added column.
    """

    out = df.copy()

    # Normalize target 'time' column to first-of-month
    out[incoming_month] = pd.to_datetime(out[incoming_month],dayfirst=True).dt.to_period("M").dt.to_timestamp()
    out = out[out[incoming_month] >= "2015-01-01"]
    # Load the incoming data
    incoming = pd.read_csv(csv_path)
    
    # Normalize and select needed columns
    incoming = incoming.copy()
    incoming["time"] = (
        pd.to_datetime(incoming["time"])
        .dt.to_period("M")
        .dt.to_timestamp()
    )
    
    # Keep only time + data columns and rename
    out = out[[incoming_month, incoming_data]].rename(columns={incoming_data: new_column_name, incoming_month: 'time'})

    # Merge based on month
    incoming = incoming.merge(out, on="time", how="left")

    return incoming
```
